```python
import zipfile
```

```python
ZIP_PATH = "Files/NYC_Taxi/taxi_zones/taxi_zones.zip"
OUTPUT_PATH = "Files/NYC_Taxi/taxi_zones"
```

```python
def extract_zip(zip_path, output_path):
    with zipfile.ZipFile(f"/lakehouse/default/{zip_path}", "r") as zip_file:
        zip_file.extractall(f"/lakehouse/default/{output_path}")
```

```python
def delete_zip(zip_path):
    notebookutils.fs.rm(zip_path)
```

```python
def process(zip_path, output_path):
    extract_zip(zip_path, output_path)
    delete_zip(zip_path)
```

```python
process(ZIP_PATH, OUTPUT_PATH)
```
