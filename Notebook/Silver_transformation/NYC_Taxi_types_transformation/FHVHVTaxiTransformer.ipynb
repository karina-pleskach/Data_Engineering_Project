```python
from pyspark.sql.functions import *
```

```python
class FHVHVTaxiTransformer(GeneralTaxiTransformer):

    TABLE_NAME = "fhvhv_tripdata"

    ADDITIONAL_SCHEMA = {
        "trip_distance": {"type": "double", "nullable": True},
        "fare_amount": {"type": "double", "nullable": True},
        "tip_amount": {"type": "double", "nullable": True},
        "total_amount": {"type": "double", "nullable": True},
        "tolls": {"type": "double", "nullable": True},
        "bcf": {"type": "double", "nullable": True},
        "sales_tax": {"type": "double", "nullable": True},
        "congestion_surcharge": {"type": "double", "nullable": True},
        "airport_fee": {"type": "double", "nullable": True},
        "cbd_congestion_fee": {"type": "double", "nullable": True}
    }
    ADDITIONAL_RESULT_COLUMNS = ["trip_distance", "fare_amount", "total_amount", "tip_amount"]
    ADDITIONAL_COLUMN_MAPPING = {
        "trip_miles": "trip_distance",
        "base_passenger_fare": "fare_amount",
        "tips": "tip_amount"
    }

    TAXI_VALUE_RANGES = {
        "trip_distance": (0, None),
        "fare_amount": (0, None),
        "total_amount": (0, None),
        "tip_amount": (0, None),
        "tolls": (0, None),
        "bcf": (0, None),
        "sales_tax": (0, None),
        "congestion_surcharge": (0, None),
        "airport_fee": (0, None),
        "cbd_congestion_fee": (0, None)
    }

    def __init__(self):
        super().__init__()
    
    def calculate_total(self, df):
        return df.withColumn("total_amount",
            col("fare_amount") + col("tolls") + col("bcf") + col("sales_tax") + col("congestion_surcharge") + col("airport_fee") + col("tip_amount") + col("cbd_congestion_fee")
        )
```
