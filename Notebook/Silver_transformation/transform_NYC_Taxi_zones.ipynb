```python
from pyspark.sql.functions import *
```

```python
def change_unknown_zones(df):
    special_zones = ["Unknown", "Outside of NYC"]
    df = df.withColumn("Borough", when(col("Zone").isin(special_zones), col("Zone")).otherwise(col("Borough")))
    df = df.withColumn("Zone", when(col("Zone").isin(special_zones), None).otherwise(col("Zone")))
    return df
```

```python
def change_airport_service_zones(df):
    return df.withColumn("service_zone", when(lower(col("Zone")).contains("airport"), "Airports").otherwise(col("service_zone")))
```

```python
def transform_taxi_zones(df):
    df = df.replace("N/A", None)
    df = change_unknown_zones(df)
    df = change_airport_service_zones(df)
    return df
```

```python
def validate_taxi_zones(df):
    validate_no_nulls(df, "LocationID")
    validate_unique(df, "LocationID")
    validate_no_empty_strings(df, ["Borough", "Zone", "service_zone"])
    return True
```

```python
def process():
    df = read_bronze_table("NYC_Taxi_zones")
    df = transform_taxi_zones(df)
    if validate_taxi_zones(df):
        save_silver_table(df, "NYC_Taxi_zones")
    return df
```
