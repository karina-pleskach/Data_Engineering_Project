```python
def transform_gdp_values(df):
    df = select_columns(df, ["date", "value"])
    df = rename_columns(df, {"date": "year", "value": "gdp_value"})
    return df
```

```python
def process():
    df = read_bronze_table("USA_GDP_values")
    df = transform_gdp_values(df)
    save_silver_table(df, "USA_GDP_values")
    return df
```
