```python
from pyspark.sql.functions import *
```

```python
def validate_no_nulls(df, column):
    if df.filter(col(column).isNull()).count() > 0:
        raise ValueError(f"{column} contains NULL values")
```

```python
def validate_unique(df, column):
    if df.groupBy(column).count().filter(col("count") > 1).count() > 0:
        raise ValueError(f"{column} contains duplicates")
```

```python
def validate_no_empty_strings(df, columns):
    for column in columns:
        if df.filter(trim(col(column)) == "").count() > 0:
            raise ValueError(f"{column} contains empty strings")
```
