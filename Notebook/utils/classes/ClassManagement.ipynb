```python
import inspect
```

```python
class ClassManagement:

    @staticmethod
    def get_all_subclasses(cls):
        subclasses = cls.__subclasses__()
        return subclasses + [subclass for child in subclasses for subclass in ClassManagement.get_all_subclasses(child)]

    @staticmethod
    def get_all_subclass_instances(cls):
        return [c() for c in ClassManagement.get_not_abstract_subclasses(cls)]

    @staticmethod
    def get_all_instances(clss):
        return [cls() for cls in clss]

    @staticmethod
    def get_not_abstract_subclasses(cls):
        return [subclass for subclass in ClassManagement.get_all_subclasses(cls) if not inspect.isabstract(subclass)]
```
