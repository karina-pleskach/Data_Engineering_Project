```python
import inspect
```

```python
class ClassManagement:

    @staticmethod
    def get_all_subclasses(cls):
        subclasses = cls.__subclasses__()
        return subclasses + [subclass for child in subclasses for subclass in ClassManagement.get_all_subclasses(child)]

    @staticmethod
    def get_all_instances(clss):
        return [cls() for cls in clss if not inspect.isabstract(cls)]
```
