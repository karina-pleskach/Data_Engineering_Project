```python
from pyspark.sql.functions import *
```

```python
class ZoneTable:

    TABLE_NAME = "nyc_taxi_zones"

    def __init__(self, table_name = None):
        self.table_name = table_name or self.TABLE_NAME
        self.df = read_table(self.table_name, "Silver_lakehouse")

    def get_special_location_ids(self):
        special_zones = {}
        for borough in ["UNKNOWN", "OUTSIDE OF NYC"]:
            zone = self.df.filter(upper(trim(col("Borough"))) == borough).select("LocationID").first()
            if zone is None:
                raise ValueError(f"Could not find {borough} zone")
            special_zones[borough] = zone["LocationID"]
        return special_zones["UNKNOWN"], special_zones["OUTSIDE OF NYC"]

    def get_valid_location_ids(self):
        return [row["LocationID"] for row in self.df.select("LocationID").collect()]
```
