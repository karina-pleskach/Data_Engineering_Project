```python
from pyspark.sql.functions import *
from pyspark.sql import Window
```

```python
def select_columns(df, columns):
    return df.select(*columns)
```

```python
def rename_column(df, old_name, new_name):
    return df.withColumnRenamed(old_name, new_name)
```

```python
def rename_columns(df, column_mapping):
    for old_name, new_name in column_mapping.items():
        df = df.withColumnRenamed(old_name, new_name)
    return df
```

```python
def add_missing_columns(df, required_columns):
    for column, data_type in required_columns.items():
        if column not in df.columns:
            df = df.withColumn(column, lit(None).cast(data_type))
    return df
```

```python
def cast_columns(df, schema):
    for column, data_type in schema.items():
        df = df.withColumn(column, col(column).cast(data_type))
    return df
```

```python
def join_dataframes(left_df, right_df, column, how="inner"):
    return left_df.join(right_df, on=column, how=how)
```

```python
def get_min_max_values(df, column):
    return df.agg(
        min(column).alias("min_value"),
        max(column).alias("max_value")
    ).first()
```

```python
def fill_forward(df, column, order_by):
    window = (Window.orderBy(order_by))
    return df.withColumn(column, last(column, ignorenulls=True).over(window))
```
