```python
import com.microsoft.spark.fabric
from com.microsoft.spark.fabric.Constants import Constants
```

```python
class DimAirQualityParameterTransformer:

    def __init__(self):
        self.source = OpenAQDailyTable()
        self.dimParameter = DimAirQualityParameter()

    def get_new_data(self):
        return self.source.df.select("parameter_id", "parameter_name", "parameter_display_name", "parameter_units").distinct()   

    def transform(self):
        new_data = self.get_new_data()
        self.dimParameter.insert_new(new_data)
```

```python
DimAirQualityParameterTransformer().transform()
```
