```python
class DimTaxiTypeTransformer:

    def __init__(self):
        self.source_classes = ClassManagement.get_not_abstract_subclasses(TaxiSilverTable)
        self.dimTaxiType = DimTaxiType()

    def get_taxi_types(self):
        return [(cls.TAXI_TYPE_ID, cls.TABLE_NAME.split("_")[0]) for cls in self.source_classes]

    def transform(self):
        taxi_types = self.get_taxi_types()
        new_df = spark.createDataFrame(taxi_types,["taxi_type_id", "taxi_type"])\
            .withColumn("taxi_type_id", F.col("taxi_type_id").cast("smallint"))
        self.dimTaxiType.insert_new(new_df)
```
