```python
class GoldTable(Table, ABC):

    STORAGE_NAME = "Gold_warehouse"

    def read(self):
        return spark.read.synapsesql(f"{self.STORAGE_NAME}.dbo.{self.TABLE_NAME}")
    
    def append(self):
        self.df.write.mode("append").synapsesql(f"{self.STORAGE_NAME}.dbo.{self.TABLE_NAME}")

    
```
