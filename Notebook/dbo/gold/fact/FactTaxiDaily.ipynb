```python
class FactTaxiDaily(GoldTable, TableWithPKS, TableWithDate):

    TABLE_NAME = "FactTaxiDaily"
    PKS = ["date", "pickup_location_id", "dropoff_location_id", "taxi_type_id"]
    DATE_COLUMN = "date"

    def loaded_months(self, taxi_type_id):
        return self.filter_rows("taxi_type_id", taxi_type_id, "=")\
                .enrich_by_year_month()\
                .available_year_months()
        
    def new_months(self, silver_df, taxi_type_id):
        available_months = silver_df.loaded_months()
        loaded_months = self.loaded_months(taxi_type_id)
        return available_months.join_on_year_month(loaded_months.df, how="left_anti")
```
