```python
from abc import ABC, abstractmethod
```

```python
class TableWithPKS(ABC):

    @property
    @abstractmethod
    def PKS(self) -> list[str]:
        pass
    
    @abstractmethod
    def append(self):
        pass

    def unique_pk_fields(self):
        new_df =  self.df.select(*self.PKS).distinct()
        return type(self)(new_df)
        
    def new_rows(self, new_df):
        current_rows = self.unique_pk_fields().df
        new_rows = new_df.join(current_rows, on=self.PKS, how="left_anti")
        return type(self)(new_rows)

    def insert_new(self, new_df):
        self.new_rows(new_df).append()
```
